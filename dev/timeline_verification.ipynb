# AI Paradigm Evolution & Keyword Timeline (1986–2028): End-to-End Verification Pipeline
**BLUF:** This executable verification notebook (`timeline_pipeline.py` $\rightarrow$ `timeline_verification.ipynb`) deterministically builds and audits the Single Source of Truth (SSOT) datasets in `cache_timeline/`, verifies all 8 publication visualisations across `figures_timeline/` and `figures_v5/`, audits both standalone reports (`2026 11 04 AI_timeline report.md` and `2026 11 04 AI_usage_level_keyword_quiz.md`), and executes all 8 Programmatic Validation Gates (`Gate 1`–`Gate 8`).

```python
from pathlib import Path
import numpy as np
import pandas as pd
from PIL import Image

from timeline_data_and_gates import (
    WORKSPACE_DIR,
    CACHE_DIR,
    FIG_DIR,
    FIG_V5_DIR,
    build_all_timeline_datasets,
    run_gates_1_to_7,
    run_gate_8_report_verification,
)

REPORT_MAIN = WORKSPACE_DIR / "2026 11 04 AI_timeline report.md"
REPORT_QUIZ = WORKSPACE_DIR / "2026 11 04 AI_usage_level_keyword_quiz.md"

print(f"Active Workspace Root: {WORKSPACE_DIR}")
```

## 1. Build & Audit Single Source of Truth (SSOT) Datasets (`cache_timeline/*.csv`)
**Dataset Schema Verification:** Generates and verifies all 5 tabular SSOT CSV files (`timeline_generations.csv`, `timeline_telemetry.csv`, `timeline_telemetry_anchors.csv`, `timeline_lines.csv`, and `timeline_events.csv`) with zero `NaN`/null entries.

```python
dfs = build_all_timeline_datasets()

for name, df in dfs.items():
    null_cnt = int(df.isna().sum().sum())
    print(f"[SSOT DATASET] {name:18s} | shape={str(df.shape):12s} | nulls={null_cnt}")
    assert null_cnt == 0, f"Dataset {name} contains {null_cnt} unexpected nulls!"
```

## 2. Macro-Generation (`GEN 0`–`GEN 6`) & 6-Category Taxonomy Summary
**Chronological & Taxonomic Audit:** Summarises the 7 chronological eras (`GEN 0` through `GEN 6`), the 6 orthogonal event categories (`PAPERS`, `RELEASES`, `INCIDENTS`, `ECON`, `COURSES`, `BETS`), and the 16 horizontal continuity tracks (`7` Company Product Lines + `9` Tech-Transfer TRL Lines).

```python
df_gen = dfs["generations"]
df_ev = dfs["events"]
df_lines = dfs["lines"]

print("=== 7 CHRONOLOGICAL GENERATIONS (GEN 0 .. GEN 6) ===")
print(
    df_gen[
        ["gen_id", "era_name", "start_ym", "end_ym", "users_span", "money_span", "tokens_span", "event_count"]
    ].to_string(index=False)
)

print("\n=== EVENT DISTRIBUTION BY GENERATION ===")
gen_summary = (
    df_ev.groupby("gen_id", sort=False)
    .agg(
        total_events=("event_id", "count"),
        mini_flagships=("mini_include", "sum"),
        mean_salience=("salience_score", lambda s: round(float(s.mean()), 3)),
        max_font_mega=("font_size_mega", "max"),
    )
    .reset_index()
)
print(gen_summary.to_string(index=False))

print("\n=== EVENT DISTRIBUTION BY 6 ORTHOGONAL CATEGORIES ===")
cat_summary = (
    df_ev.groupby(["category_key", "category"], as_index=False)
    .agg(
        event_count=("event_id", "count"),
        mini_count=("mini_include", "sum"),
        mean_salience=("salience_score", lambda s: round(float(s.mean()), 3)),
        median_salience=("salience_score", lambda s: round(float(s.median()), 3)),
    )
    .sort_values("event_count", ascending=False)
)
print(cat_summary.to_string(index=False))
```

## 3. CITRA Section 4.10 Animation Mathematical Frame Breakdown (`Plot 6`)
**Animation Telemetry Verification:** Audits `figures_timeline/fig_timeline_p6_timeline_anim.gif` against the 6 required CITRA pre-render mathematical frame equations (`96` frames @ `4 FPS` = `24.00 seconds`).

```python
total_intervals = 96
SKIP_INTERVALS = 1
MAX_FRAMES_TO_RENDER = 96
ANIMATION_FPS = 4
raw_frames = total_intervals // SKIP_INTERVALS
actual_frames = min(raw_frames, MAX_FRAMES_TO_RENDER)
expected_duration_sec = actual_frames / ANIMATION_FPS

print("=" * 78)
print("CITRA ANIMATION PRE-RENDER MATHEMATICAL FRAME BREAKDOWN (Plot 6)")
print(f"1. Total dataset intervals:   {total_intervals}")
print(f"2. Skip interval stride:      SKIP_INTERVALS = {SKIP_INTERVALS}")
print(f"3. Resulting frame count:     {total_intervals} // {SKIP_INTERVALS} = {raw_frames}")
print(f"4. Frame cap applied:         min({raw_frames}, MAX_FRAMES_TO_RENDER={MAX_FRAMES_TO_RENDER}) = {actual_frames}")
print(f"5. Animation FPS:             ANIMATION_FPS = {ANIMATION_FPS} frames/sec")
print(f"6. Expected output duration:  {actual_frames} / {ANIMATION_FPS} = {expected_duration_sec:.2f} seconds")
print("=" * 78)

gif_path = FIG_DIR / "fig_timeline_p6_timeline_anim.gif"
with Image.open(gif_path) as im_gif:
    n_frames_gif = getattr(im_gif, "n_frames", 1)
    w_gif, h_gif = im_gif.size
print(f"[VERIFIED GIF] {gif_path.name} | dimensions={w_gif}x{h_gif}px | n_frames={n_frames_gif} | size={gif_path.stat().st_size:,} bytes")
assert n_frames_gif == 96, f"Expected 96 frames in Plot 6 GIF, got {n_frames_gif}"
```

## 4. Visual Asset & Standalone Report Integrity Audit (`Plots 1–8` & Reports)
**File & Resolution Verification:** Confirms pixel dimensions and byte sizes of all 9 visual files (`Plots 1–6` in `figures_timeline/` and `Plots 7–8` + companion self-diagnosis chart in `figures_v5/`) and both Markdown reports.

```python
visual_files = [
    FIG_DIR / "fig_timeline_p1_mega_poster.png",
    FIG_DIR / "fig_timeline_p2_mini.png",
    FIG_DIR / "fig_timeline_p3_macro_telemetry.png",
    FIG_DIR / "fig_timeline_p4_salience_money_jointplot.png",
    FIG_DIR / "fig_timeline_p5_category_raster_and_trl_lag.png",
    FIG_DIR / "fig_timeline_p6_timeline_anim.gif",
    FIG_V5_DIR / "fig_v5_quiz_p1_blind_and_reveal_board.png",
    FIG_V5_DIR / "fig_v5_quiz_p2_6x6_era_level_atlas.png",
    FIG_V5_DIR / "fig_v5_current_gen_self_diagnosis.png",
]

for vf in visual_files:
    assert vf.exists() and vf.stat().st_size > 0, f"Missing or empty visual asset: {vf}"
    with Image.open(vf) as img:
        w, h = img.size
        nf = getattr(img, "n_frames", 1)
    print(f"[ASSET OK] {vf.relative_to(WORKSPACE_DIR)} | {w}x{h}px | frames={nf} | {vf.stat().st_size:,} bytes")

for rp in [REPORT_MAIN, REPORT_QUIZ]:
    assert rp.exists() and rp.stat().st_size > 0, f"Missing or empty report: {rp}"
    lines_cnt = len(rp.read_text(encoding="utf-8").splitlines())
    print(f"[REPORT OK] {rp.name} | {lines_cnt:,} lines | {rp.stat().st_size:,} bytes")
```
